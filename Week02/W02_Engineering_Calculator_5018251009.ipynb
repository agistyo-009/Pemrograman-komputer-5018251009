```python
# ==============================================================================
# TASK 1: Speed Unit Conversion (km/h to m/s)
# ==============================================================================
print("=== TASK 1: SPEED UNIT CONVERSION ===")
speed_kmh = float(input("Enter speed [km/h]: "))
speed_ms = speed_kmh / 3.6
print(f"Speed = {speed_ms:.2f} m/s\n")
```

```python
# ==============================================================================
# TASK 2: Power Unit Conversion (kW to W)
# ==============================================================================
print("=== TASK 2: POWER UNIT CONVERSION ===")
power_kw = float(input("Enter power [kW]: "))
power_w = power_kw * 1000.0
print(f"Power = {power_w:.2f} W\n")
```

```python
# ==============================================================================
# TASK 3: Electrical Power and Energy Calculator
# ==============================================================================
print("=== TASK 3: ELECTRICAL POWER AND ENERGY CALCULATOR ===")
voltage = float(input("Enter voltage [V]: "))
current = float(input("Enter current [A]: "))
operating_time = float(input("Enter operating time [h]: "))

power = voltage * current
energy = power * operating_time

print("\n===== ELECTRICAL CALCULATION =====")
print(f"Voltage        = {voltage:.2f} V")
print(f"Current        = {current:.2f} A")
print(f"Power          = {power:.2f} W")
print(f"Operating Time = {operating_time:.2f} hour")
print(f"Energy         = {energy:.2f} Wh\n")
```

```python
# ==============================================================================
# TASK 4: Comparison and Logical Operators
# ==============================================================================
print("=== TASK 4: COMPARISON AND LOGICAL OPERATORS ===")
power_limit = 200.0
voltage_limit = 24.0

is_high_power = power > power_limit
is_high_voltage = voltage > voltage_limit

print(f"High power      : {is_high_power}")
print(f"High voltage    : {is_high_voltage}")
print(f"Both conditions : {is_high_power and is_high_voltage}")
print(f"Any condition   : {is_high_power or is_high_voltage}\n")
```

## Short Reflection

1. **Apa perbedaan `int` dan `float`?**
   - `int` (*integer*) adalah tipe data untuk bilangan bulat tanpa desimal (contoh: `12`, `5`, `-3`).
   - `float` (*floating-point*) adalah tipe data untuk bilangan real/pecahan yang memiliki titik desimal (contoh: `12.5`, `3.6`, `0.0`).

2. **Mengapa hasil `input()` perlu dikonversi menggunakan `float()` untuk perhitungan numerik?**
   - Fungsi `input()` secara default selalu mengembalikan nilai bernilai tipe data teks (*string* / `str`). Jika string dikenakan operator matematika seperti `+` atau `*`, Python akan melakukan penggabungan teks alih-alih perhitungan numerik. Konversi `float()` mengubah teks tersebut menjadi angka desimal agar dapat dihitung matematis.

3. **Apa perbedaan operator `=` dan `==` dalam Python?**
   - `=` adalah **operator penugasan** (*assignment operator*) yang digunakan untuk memasukkan/menyimpan nilai ke dalam suatu variabel.
   - `==` adalah **operator perbandingan** (*comparison operator*) yang digunakan untuk mengecek apakah dua nilai bernilai sama, menghasilkan boolean (`True` atau `False`).

4. **Apa fungsi operator `and` dan `or`?**
   - `and`: Mengembalikan nilai `True` **hanya jika kedua kondisi bernilai True**. Jika salah satu atau keduanya False, hasilnya False.
   - `or`: Mengembalikan nilai `True` **jika salah satu atau kedua kondisi bernilai True**. Hanya akan bernilai False jika kedua kondisi bernilai False.

5. **Apa satu kesalahan yang Anda temui saat mengerjakan tugas ini, dan bagaimana Anda membaikinya?**
   - *Contoh:* Kesalahan yang sempat ditemui adalah lupa mengonversi tipe data hasil `input()`, sehingga program memunculkan error `TypeError` saat dilakukan operasi pembagian `speed_kmh / 3.6`. Kesalahan diperbaiki dengan membungkus fungsi `input()` ke dalam `float()`, yaitu `float(input(...))`.
