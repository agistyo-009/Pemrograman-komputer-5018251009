```python

import numpy as np
from scipy.optimize import root_scalar

def f(x):
    return x**2 - 16

hasil = root_scalar(f, bracket=[0, 6], method='brentq')

print("=== 1. Persamaan Nonlinear ===")
print("Akar persamaan:", hasil.root)
print("Konvergensi:", hasil.converged)
print("Residual:", f(hasil.root))


# ==========================================
# 2. MENYELESAIKAN SISTEM PERSAMAAN LINEAR
# 3x + y = 7
# x + 2y = 5
# ==========================================

A = np.array([[3, 1],
              [1, 2]])

b = np.array([7, 5])

solusi = np.linalg.solve(A, b)

print("\n=== 2. Sistem Persamaan Linear ===")
print("Nilai x:", solusi[0])
print("Nilai y:", solusi[1])


# ==========================================
# 3. MENGHITUNG RESIDUAL KEDUA HASIL
# ==========================================

residual_1 = f(hasil.root)
residual_2 = A @ solusi - b

print("\n=== 3. Perhitungan Residual ===")
print("Residual nonlinear:", residual_1)
print("Residual linear:", residual_2)


# ==========================================
# 4. MENGGANTI MATRIKS MENJADI SINGULAR
# ==========================================

A_singular = np.array([[3, 1],
                       [6, 2]])

print("\n=== 4. Matriks Singular ===")
print("Determinan:", np.linalg.det(A_singular))

try:
    solusi_singular = np.linalg.solve(A_singular, b)
    print("Solusi:", solusi_singular)

except np.linalg.LinAlgError as error:
    print("Pesan kesalahan:", error)
```

#Validasi Solusi Numerik
##Solusi numerik dapat dipercaya apabila:
1. Metode berhasil mencapai konvergensi.
2. Nilai residual mendekati nol.
3. Matriks tidak singular.
4. Hasil perhitungan stabil.
5. Hasil sesuai dengan kondisi permasalahan.

```python

import numpy as np

# ==========================================
# 1. MENYELESAIKAN PERSAMAAN NONLINEAR
# x^2 - 16 = 0 pada interval [0, 6]
# Menggunakan metode Bisection
# ==========================================

def f(x):
    return x**2 - 16

a = 0
b = 6
toleransi = 1e-10

while abs(b - a) > toleransi:
    c = (a + b) / 2

    if f(a) * f(c) < 0:
        b = c
    else:
        a = c

akar = (a + b) / 2

print("=== 1. Persamaan Nonlinear ===")
print("Akar persamaan:", akar)
print("Residual:", f(akar))


# ==========================================
# 2. SISTEM PERSAMAAN LINEAR
# 3x + y = 7
# x + 2y = 5
# ==========================================

A = np.array([[3, 1],
              [1, 2]])

B = np.array([7, 5])

solusi = np.linalg.solve(A, B)

print("\n=== 2. Sistem Persamaan Linear ===")
print("Nilai x:", solusi[0])
print("Nilai y:", solusi[1])


# ==========================================
# 3. MENGHITUNG RESIDUAL KEDUA HASIL
# ==========================================

residual_1 = f(akar)
residual_2 = A @ solusi - B

print("\n=== 3. Perhitungan Residual ===")
print("Residual nonlinear:", residual_1)
print("Residual linear:", residual_2)


# ==========================================
# 4. MENGGANTI MATRIKS MENJADI SINGULAR
# ==========================================

A_singular = np.array([[3, 1],
                       [6, 2]])

print("\n=== 4. Matriks Singular ===")
print("Determinan:", np.linalg.det(A_singular))

try:
    solusi_singular = np.linalg.solve(A_singular, B)
    print("Solusi:", solusi_singular)

except np.linalg.LinAlgError as error:
    print("Pesan kesalahan:", error)
```
