```python

# Program Sensor Log Analyzer

sensor = [25, 30, 45, 20, -5, 60, 35, 40, 15, 70,
          28, 32, 55, 18, 42, -10, 65, 22, 38, 50]

batas_alarm = 50

jumlah_data = 0
total = 0
jumlah_alarm = 0

nilai_min = None
nilai_max = None

print("=== SENSOR LOG ANALYZER ===")
print("Batas alarm:", batas_alarm)
print()

# Menampilkan nomor sampel, nilai, dan status
for i in range(len(sensor)):
    nilai = sensor[i]

    if nilai < 0:
        print("Sampel", i + 1, ":", nilai, "- DATA TIDAK VALID")
        continue

    if nilai >= batas_alarm:
        status = "ALARM"
        jumlah_alarm = jumlah_alarm + 1
    else:
        status = "NORMAL"

    print("Sampel", i + 1, ":", nilai, "-", status)

    # Menghitung rata-rata
    total = total + nilai
    jumlah_data = jumlah_data + 1

    # Mencari nilai minimum
    if nilai_min is None or nilai < nilai_min:
        nilai_min = nilai

    # Mencari nilai maksimum
    if nilai_max is None or nilai > nilai_max:
        nilai_max = nilai

# Menampilkan hasil analisis
print("\n=== HASIL ANALISIS ===")

if jumlah_data > 0:
    rata_rata = total / jumlah_data

    print("Jumlah data valid:", jumlah_data)
    print("Rata-rata sensor:", round(rata_rata, 2))
    print("Jumlah alarm:", jumlah_alarm)
    print("Nilai minimum:", nilai_min)
    print("Nilai maksimum:", nilai_max)
else:
    print("Tidak ada data sensor yang valid.")
```
